```python
import numpy as np
import pandas as pd
arr=np.array([10,20,30,40,50])
print("array:",arr)
print("Mean is:",arr.mean())
print("Reshaped 2D:\n",arr.reshape(5,1))
print("Sliced from index 1,2,3:",arr[1:4])
data ={
    'Name' : ['Anjali','Akshatha','Anagha','Ashiq'],
    "Marks" : [86,76,66,56],
    "Branch" : ['CSE','CSE','CSE','CSE']
}
df = pd.DataFrame(data)
print("\n First 2 Rows:\n",df.iloc[0:4])
print("\n Marks Column \n",df['Marks'])
print("\n Rows where Marks>70:\n",df[df['Marks']>70])
```
